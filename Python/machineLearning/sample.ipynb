```python
import numpy as np
from sklearn.linear_model import LinearRegression
```

```python
# col: study hours
X=np.array([[12],[5],[15],[4]])

# col: internal marks
y=np.array([78, 45, 88, 39])

# initializing the linear regression model
model=LinearRegression()

#  training the model
model.fit(X,y)

ans=model.predict([[8]])
print(ans)

# to change 2d array to 1d
X1=X.ravel()

#  calculating correlation: as two columns can be used to fill the missing data
np.corrcoef(X1,y)
```

```python
#  co: attendance
att=np.array([[85],[62],[91],[55]])
y=np.array([78, 45, 88, 39])

model=LinearRegression()
model.fit(att,y)

ans=model.predict([[74]])
print(ans)

X1=X.ravel()
np.corrcoef(X1,y)
```

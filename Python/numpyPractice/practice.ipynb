```python
marks=marks+5
```

Indexing

```python
# accessed using both positive and negative indexing
marks[3]
```

Traversing

```python
for i in range(len(marks)):
    print(marks[i])
```

Slicing

```python
marks[2:]
```

2D Array

```python
marks=np.array([[85,78,90],[92,85,98],[76,82,80]])
print(marks)
```

2D Slicing

```python
marks[0:2]
marks[0:2,0:2]
```

```python
print(marks[0,2])
print(marks[1,0:2])
print(marks[1][0:2])
```

```python
marks[1][2]
```

```python
print(marks[1:,1])
print(marks[1:][1])
```

Concatenation

```python
b1=np.array([56,78,89])
b2=np.array([65,76,98])

comb1=np.concatenate((b1,b2))
print(comb1)
```

2D Concatenation 

```python
b1=np.array([[56,78,89],[34,45,67]])
b2=np.array([[65,76,98],[89,98,78]])

comb2=np.concatenate((b1,b2))
print(comb2)
```

```python
b1=np.array([[56,78,89],[34,45,67]])
b2=np.array([[65,76,98],[89,98,78]])

comb2=np.concatenate((b1,b2),axis=1)
print(comb2)
```

Split - divides array in equal division

```python
np.split(comb1,3)
```

Search

```python
print(marks)
print(np.where(marks==98))
print(np.where(marks>=80))
print(marks[np.where(marks>=80)])
```

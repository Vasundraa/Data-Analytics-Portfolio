```python
import pandas as pd
```

```python
data = {
    "Student_ID": [101, 102, 103, 104, 105],
    "Name": ["Arun", "Priya", "Karthik", "Meena", "Rahul"],
    "Department": ["CSE", "IT", "CSE", "ECE", "IT"],
    "Attendance": [82, 91, 76, 68, 88],
    "Python": [85, 92, 76, 65, 88],
    "SQL": [78, 88, 82, 70, 91],
    "Excel": [90, 95, 80, 72, 86]
}
```

Creating Data Frame - Tabular Structure

```python
df=pd.DataFrame(data)
print(df)
```

Accessing

```python
#accessing a column
print(df['Python'])

print(df['Python'].dtype)
```

```python
#accessing specific columns
print(df[['Name','Python','SQL','Excel']])
```

```python
#accessing specific row
print(df.iloc[0])
```

```python
#accessing a particular column in row access
print(df.iloc[0,3])
```

```python
#accessing multiple rows
print(df.iloc[[1,2]])
```

```python
#accessing multiple rows and multiple columns
print(df.iloc[0:4,[1,2]])
```

```python
print(df.loc[0:3,['Name']])
```

```python
#accessing based on condition
#students with atleast 75 attendance
print(df.loc[df['Attendance']>=75,['Name','Attendance']])
```

```python
#students with atleast 75 attendance and atleast 80 in python
res=df[
    (df['Attendance']>=75) &
    (df['Python']>=80)
]

print(res['Name','Attendance','Python'])
```

```python
#Average of Python, SQL and Excel
print((df['Python']+df['SQL']+df['Excel'])//3)
```

```python
#adding new column
df["Average"]=(df['Python']+df['SQL']+df['Excel'])//3
print(df)
```

```python
#adding another column based on conditions applied to existing column
df['Result']=df['Average'].apply( lambda x: "Pass" if x>=50 else "Fail")
print(df)
```

Sorting

```python
df.sort_values(by='Average')
```

```python
df.sort_values(by='Average',ascending=False)
```

Group By

```python
#average of python in each dept
df.groupby('Department')['Python'].mean()
```

```python
#no. of students per dept
df.groupby('Department')['Student_ID'].count()
```

```python
#highest python mark by dept
df.groupby('Department')['Python'].max()
```

```python
#group by with multiple columns
df.groupby('Department').agg(
    {
        'Student_ID':'count',
        'Python':'mean',
        'SQL':'mean',
        'Attendance':'mean'
    }
)
```
